### Обучающая и тестовая выборки

- В качестве обучающей выборки будем использовать наборы, полученные при помощи train_test_split из scikit-learn.
- Конечно, алгоритм может работать с многомерным пространством, но для удобства демонстраиции алгоритма мы используем двухмерный набор данных.
- Если разные признаки имеют сильно отличающиеся диапазоны значений, то применяют масштабирование исходных данных.

```python
# Установим необходимые библиотеки (раскоментировать)
#%pip install numpy pandas sklearn
```

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

```python
data_full = pd.read_csv('data/dataset.csv', sep=",")
data_full
```

Проверим датасет на пустые значения. 

```python
# проверим есть ли пропущенные значения
data_full.isnull().sum()
```

Заметим, что NObeyesdad является категориальным признаком, рассмотрим уникальные значения для него:

```python
np.unique(data_full[['NObeyesdad']])
```

Приведем признак к нумерическому типу.

```python
def target_to_regr(target: str) -> float:
    mapping = {
        'Insufficient_Weight': 0,
        'Normal_Weight': 0,
        'Overweight_Level_I': 1,
        'Overweight_Level_II': 1,
        'Obesity_Type_I': 1,
        'Obesity_Type_II': 1,
        'Obesity_Type_III': 1
    }
    return mapping.get(target, 2)  # Возвращает 2, если категория не найдена
```

```python
# Для начала, преобразуем целевой признак в категориальный для примера
data_full['Obesity'] = \
data_full.apply(lambda row: target_to_regr(row['NObeyesdad']),axis=1)
data_full=data_full.drop(columns='NObeyesdad')
data_full
```

Создадим регрессионный целевой признак BMI.

```python
def class_to_regr(height:float, weight: float) -> float:
    result = weight/(np.square(height))
    return result 
```

```python
data_full['BMI'] = \
data_full.apply(lambda row: class_to_regr(row['Height'],row['Weight']),axis=1)
```

Закодируем остальные строковые показатели при помощи простого LabelEncoder.

```python
categorical_features = [
    'Gender', 'family_history_with_overweight', 'FAVC', 'CAEC', 
    'SMOKE', 'SCC', 'CALC', 'MTRANS', 'Obesity'
]

# Инициализация LabelEncoder
label_encoders = {}
```

```python
# Перекодировка категориальных признаков
for feature in categorical_features:
    if feature in data_full.columns:
        le = LabelEncoder()
        data_full[feature] = le.fit_transform(data_full[feature])
        label_encoders[feature] = le
```

Разделим выборку на тестовую и тренировочную.

```python
X = data_full.drop(columns=['Obesity'])  # Let's say column 'target' is the target\class\label column
y = data_full['Obesity']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,
                                                    train_size=0.5, 
                                                    random_state=42)
```

```python
data_train_to_csv=X_train
data_train_to_csv['Obesity']=y_train
data_train_to_csv
```

```python
data_test_to_csv=X_test
data_test_to_csv['Obesity']=y_test
data_test_to_csv
```

Для второй тестовой выборки возьмем все данные датасета.

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.99,
                                                    train_size=0.001, 
                                                    random_state=42)
data_test_to_csv2=X_test
data_test_to_csv2['Obesity']=y_test
data_test_to_csv2
```

Запишем полученные датасеты в .csv и запишем в новые переменные (для чистоты эксперимента).

```python
pd.DataFrame.to_csv(data_test_to_csv, './data/obesity_test.csv', sep=",", index=False)
pd.DataFrame.to_csv(data_train_to_csv, './data/obesity_train.csv', sep=",", index=False)
pd.DataFrame.to_csv(data_test_to_csv2, './data/obesity_test_full.csv', sep=",", index=False)
```
